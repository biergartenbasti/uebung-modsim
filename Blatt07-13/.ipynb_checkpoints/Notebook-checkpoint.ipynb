Im Folgenden wird die Diffusion eines Metallatoms auf simplen Mg- oder Li- 001-Metalloberflächen modelliert. Die benötigten Dateien liegen in den Ordnern Li/ und Mg/ bereit, es sollen nur die POSCAR-Strukturdateien genutzt werden. Die anderen Dateien spielen für die Berechnung keine Rolle.
Bei den Strukturdateien handelt es sich um periodische Superzellen der Größe 4x4x3 mit einem Vacuum von 20 Angstrom oberhalb der Struktur. Als Referenzstruktur der Superzelle dient hier die primitive Zelle, die einer Größe von 1x1x1 entspricht.
